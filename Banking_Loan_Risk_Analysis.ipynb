Missing Value Handle

```python

print(df.isnull().sum())
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values handled successfully!")

```

Remove Duplicate Records

```python
df = df.drop_duplicates()

print("Dataset shape after cleaning:", df.shape)
```

Check Clean Dataset

```python
df.isnull().sum()
```

Calculate 8 Important KPIs

```python
total_applicants = len(df)

default_cases = df['loan_status'].sum()

default_rate = (default_cases / total_applicants) * 100

average_loan = df['loan_amnt'].mean()

average_income = df['person_income'].mean()

average_interest = df['loan_int_rate'].mean()

average_age = df['person_age'].mean()

total_loan_amount = df['loan_amnt'].sum()

print("BANKING AND LOAN RISK ANALYSIS")
print("--------------------------------")
print("Total Applicants:", total_applicants)
print("Default Cases:", default_cases)
print("Loan Default Rate:", round(default_rate, 2), "%")
print("Average Loan Amount:", round(average_loan, 2))
print("Average Annual Income:", round(average_income, 2))
print("Average Interest Rate:", round(average_interest, 2), "%")
print("Average Applicant Age:", round(average_age, 2))
print("Total Loan Amount:", round(total_loan_amount, 2))
df
```

Loan Default Analysis

```python
print(df['loan_status'].value_counts())

sns.countplot(data=df, x='loan_status')
plt.title("Loan Default Distribution")
plt.xlabel("Loan Status (0 = Non-Default, 1 = Default)")
plt.ylabel("Number of Applicants")
plt.show()
```

Loan Amount Distribution Graph

```python
plt.figure(figsize=(8,5))
sns.histplot(df['loan_amnt'], bins=30, kde=True,
             color='purple')
plt.title("Loan Amount Distribution")
plt.xlabel("Loan Amount")
plt.ylabel("Number of Applicants")
plt.show()
```

Loan Status by Loan Intent Graph

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='loan_intent',
              hue='loan_status',
              palette=['#3498DB', '#E74C3C'])
plt.title("Loan Default by Loan Purpose")
plt.xticks(rotation=45)
plt.show()
```

Income vs Loan Amount Graph

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='person_income',
                y='loan_amnt',
                hue='loan_status',
                palette=['#16A085', '#E67E22'],
                alpha=0.6)
plt.title("Income vs Loan Amount")
plt.show()
```

Correlation Heatmap

```python
plt.figure(figsize=(10,7))
sns.heatmap(df.select_dtypes(include=np.number).corr(),
            annot=True, cmap='viridis', fmt='.2f')
plt.title("Correlation Heatmap")
plt.show()
```

Important Insights and Findings

```python
print("BANKING AND LOAN RISK ANALYSIS - KEY FINDINGS")

print("1. Total loan applications:", len(df))

print("2. Average loan amount:",
      round(df['loan_amnt'].mean(), 2))

print("3. Average customer income:",
      round(df['person_income'].mean(), 2))

print("4. Loan default percentage:",
      round(df['loan_status'].mean() * 100, 2), "%")

print("5. Most common loan purpose:",
      df['loan_intent'].mode()[0])

print("6. Most common home ownership:",
      df['person_home_ownership'].mode()[0])

print("7. Average credit history length:",
      round(df['cb_person_cred_hist_length'].mean(), 2), "years")
```
